```python
from biomni.tool import hpc_search_tools
res = hpc_search_tools('diamond blastp protein annotation')
for r in res:
    print(r.get('id'), '|', r.get('name'))
    print(r.get('description','')[:300])
    print('USAGE:', r.get('usage','')[:2000])
    print('='*80)
```
